#**Definición**
Una función f es una regla que asigna a cada número $x$ en un conjunto
$D$, un número $f(x)$. El conjunto $D$ se llama **dominio de la función** y
$f(x)$ se llama valor de la función en $x$. El conjunto de todos los valores de una función se llama **rango de la función**. El conjunto de pares ordenados $(x,f(x))$ se llama **gráfica de la función**.

##**Funciones polinomicas**
El dominio son todos los numeros reales $\mathbb{R}$ y el rango depende de cada funcion.
$$$$

$(i)$ $~~$ $f(x)=x^2-6x$
$$$$

$\operatorname{Dom}(f) =\mathbb{R} $
$$$$

Para determinar el rango
$$$$

Sea:

$$y = f(x)$$
$$$$

Completando el cuadrado
$$$$

$$ y = x^2 - 6x = (x-3)^2 - 9$$
$$$$

Como $$(x-3)^2 \ge 0$$
$$$$

para todo $x \in \mathbb{R}$, se tiene
$$$$

$$ y \ge -9$$
$$$$

Además, el mínimo se alcanza en $x=3$:
$$$$

$$f(3) = (3-3)^2 - 9 = -9$$
$$$$

$\operatorname{Ran}(f) = [-9,\infty)$
$$$$

Como ejercicio usted puede despejar $x$ en función de $y$ y verificar que se obtiene el mismo resultado.
$$$$


El dominio y el rango pueden verse claramente en la grafica.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 10, 100)
y = x**2-6*x

plt.plot(x, y)
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Gráfica de la función')
plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)
plt.plot(x,y,linewidth=1)
plt.plot(x,y,color='b')
plt.grid(color='g',linestyle='dotted', linewidth=0.5)
```

##**Funciones racionales**
El dominio tiene restricciones y el rango depende de cada función.

$(ii)$ $~~$ $f(x)=\frac{1}{x^2-1}$
$$$$

$\operatorname{Dom}(f) =\mathbb{R}-\{1,-1\}$
$$$$

Para determinar el rango
$$$$

Sea:

$$y = f(x)$$
$$$$

Entonces:


$$y = \frac{1}{x^2 - 1}$$
$$$$

$$\frac{1}{y} = x^2 - 1$$
$$$$


$$x^2 = 1 + \frac{1}{y}$$
$$$$

Por tanto:
$$$$


$$x = \pm \sqrt{1 + \tfrac{1}{y}} = \pm \sqrt{\tfrac{y+1}{y}}$$


Para  $y \neq 0$
$$$$

Para asegurar que
$$$$

$$\frac{y+1}{y} \geq 0$$
$$$$

El rango de la función es:
$$$$

$$y \in (-\infty, -1] \;\cup\; (0, \infty)$$
$$$$


Lo cual puede verse en la gráfica



```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-10, 10, 1000)
y = 1/(x**2-1)

#plt.figure(figsize=(8, 6))
plt.plot(x, y,'.')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Gráfica de la función')
plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)
plt.ylim(-2.5, 2.5)  # Establecer los límites en el eje y si es necesario
plt.grid(color='g',linestyle='dotted', linewidth=1)
```

## **Funciones Exponencial y Logaritmica**

En análisis matemático, dos funciones fundamentales son la **función exponencial** y la **función logarítmica**.  
$$$$

###**Función Exponencial**

La función exponencial se define como:

$$f(x) = a^x, \quad a > 0, \;\;\; a \neq 1$$
$$$$

### **Propiedades**:
- Su dominio es $\mathbb{R}$
- Su rango es $(0, \infty)$
- Siempre pasa por el punto $(0,1)$
- Es creciente si $a > 1$ y decreciente si $0 < a < 1$
$$$$

### **Transformaciones rígidas**

- **Traslación vertical**: $f(x) = a^x + k$  
  Mueve la gráfica hacia arriba $(k>0)$ o abajo $(k<0)$

- **Traslación horizontal**: $f(x) = a^{x-h}$  
  Mueve la gráfica hacia la derecha $(h>0)$ o izquierda $(h<0)$
$$$$


### **Ejemplos**

Funcion exponencial y trasformaciones rígidas

Considere la funcion exponencial $\;\;\; f(x)=2^x \;\;\;$ trace las graficas de $f(x), \;\;\; f(x)$  desplazada una unidad hacia arriba, $\;\;\;f(x)$ desplazada una unidad hacia abajo, $\;\;\;f(x)$ desplazada una unidad hacia la derecha, $\;\;\;f(x)$ desplazada una unidad hacia la izquierda.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2, 2,50)

#Funcion base
y =  2**x

#Transformaciones verticales

y_up=2**x+1       # 1 hacia arriba
y_down=2**x-1     # 1 hacia abajo

plt.plot(x, y, label=r"$2^x$", color='b')
plt.plot(x, y_up, label=r"$2^x+1$", color='r')
plt.plot(x, y_down, label=r"$2^x-1$", color='g')

plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Gráfica de la función exponencial y sus transformaciones verticales')
plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)
plt.plot(x,y,linewidth=1)
plt.grid(color='g',linestyle='dotted', linewidth=0.5)
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2, 2,50)

#Funcion base
y =  2**x

#Transformaciones horizontales

y_right=2**(x-1)  # 1 a la derecha
y_left=2**(x+1)   # 1 a la izquierda

plt.plot(x, y, label=r"$2^x$", color='b')
plt.plot(x, y_right, label=r"$2^{x-1}$", color='r')
plt.plot(x, y_left, label=r"$2^{x+1}$", color='g')

plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Gráfica de la función exponencial y sus transformaciones horizontales')
plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)
plt.plot(x,y,linewidth=1)
plt.grid(color='g',linestyle='dotted', linewidth=0.5)
plt.legend()
plt.show()
```

### **Funcion exponencial de base Euler**



Recuerde que la sucesión
$$$$


$${a_n}= (1 + \frac{1}{n})^n, ~~~~ n = 1,2,3,...$$
$$$$

Tiene como límite
$$$$

$$\lim_{n \to ∞}{a_n}= e$$
$$$$

Que se conoce como número de Euler
$$$$

$$e =2.71828...$$
$$$$

En anánilis es generalmente utilizada la función exponencial de base $a=e$
$$$$

$$f(x)=e^x$$
$$$$

A continuación se ilustran las transformaciones de la funcion exponencial aplicadas a esta función.
$$$$

```python
import numpy as np
import matplotlib.pyplot as plt

# --- Función Exponencial ---
x = np.linspace(-2, 2, 400)
y = np.exp(x)             # f(x) = e^x
y_up = np.exp(x) + 2       # f(x) = e^x + 2 (arriba)
y_right = np.exp(x - 1)    # f(x) = e^(x-1) (derecha)

plt.figure(figsize=(6,4))
plt.plot(x, y, label=r"$e^x$")
plt.plot(x, y_up, label=r"$e^x + 2$")
plt.plot(x, y_right, label=r"$e^{x-1}$")
plt.title("Transformaciones de la función exponencial")
plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)
plt.legend()
plt.grid(color='g',linestyle='dotted', linewidth=0.5)
plt.show()
```

### **Función Logarítmica**

La función logarítmica se define como la inversa de la exponencial:


$$g(x) = \log_a(x), \quad a > 0, \; a \neq 1$$


### **Propiedades**:
- Su dominio es $(0, \infty)$
- Su rango es $\mathbb{R}$
- Pasa por el punto $(1,0)$
- Es creciente si $a > 1$ y decreciente si $0 < a < 1$


### **Transformaciones rígidas**

- **Traslación vertical**: $g(x) = \log_a(x) + k$  
  Mueve la gráfica hacia arriba $(k>0)$ o abajo $(k<0)$

- **Traslación horizontal**: $g(x) = \log_a(x-h)$  
  Mueve la gráfica hacia la derecha $(h>0)$ o izquierda $(h<0)$

## **Función logaritmo natural**

En anánilis es generalmente utilizada la función logaritmica de base $a=e$
$$$$

$$g(x) = \log_e(x)=ln(x)$$
$$$$

llamada función logaritmo natural. A continuación se ilustran las transformaciones de la funcion logaritmo mediante aplicadas a esta función.
$$$$

```python

import numpy as np
import matplotlib.pyplot as plt

# --- Función Logarítmica ---
x = np.linspace(1.01, 5, 100)
y = np.log(x)              # g(x) = ln(x)
y_up = np.log(x) + 2        # g(x) = ln(x) + 2 (arriba)
y_right = np.log(x - 1)     # g(x) = ln(x-1) (derecha)

plt.figure(figsize=(6,4))
plt.plot(x, y, label=r"$\ln(x)$")
plt.plot(x, y_up, label=r"$\ln(x) + 2$")
plt.plot(x[x>1], y_right[x>1], label=r"$\ln(x-1)$") # dominio restringido
plt.title("Transformaciones de la función logarítmica")
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.legend()
plt.grid(color='g',linestyle='dotted', linewidth=0.5)
plt.show()
```

# **Funciones Trigonométricas**

En análisis matemático, las **funciones trigonométricas** fundamentales son el seno, coseno y tangente.  
Todas son periódicas y se utilizan ampliamente en geometría, análisis de señales y modelado de fenómenos oscilatorios.


## **Función Seno**

$$f(x) = \sin(x)$$

### **Propiedades**:

- Dominio: $\mathbb{R}$
- Rango: $[-1,1]$
- Período: $2\pi$
- Es función **impar**: $\sin(-x) = -\sin(x)$
$$$$

## **Función Coseno**

$$f(x) = \cos(x)$$


### **Propiedades**:

- Dominio: $\mathbb{R}$
- Rango: $[-1,1]$
- Período: $2\pi$
- Es función **par**: $\cos(-x) = \cos(x)$
$$$$


# **Transformaciones Rígidas**

Dada una función trigonométrica genérica $f(x)$:

- **Traslación vertical**:  
  
  $$f(x) + k$$
  
  Sube $(k>0)$ o baja $(k<0$) la gráfica.

- **Traslación horizontal**:  
  
  $$f(x-h)$$
  
  Desplaza la gráfica a la derecha $(h>0$) o izquierda $(h<0$).



```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2*np.pi, 2*np.pi, 1000)

# --- Función seno con transformaciones ---
y = np.sin(x)
y_up = np.sin(x) + 2          # arriba
y_right = np.sin(x - np.pi/2) # derecha (fase)

plt.figure(figsize=(6,4))
plt.plot(x, y, label=r"$\sin(x)$")
plt.plot(x, y_up, label=r"$\sin(x) + 2$")
plt.plot(x, y_right, label=r"$\sin(x-\pi/2)$")
plt.title("Transformaciones de la función seno")
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.ylim(-2,3.5)
plt.legend()
plt.grid(color='g',linestyle='dotted', linewidth=0.5)
plt.show()


# --- Función coseno con transformaciones ---
y = np.cos(x)
y_down = np.cos(x) - 1         # abajo
y_left = np.cos(x + np.pi/2)   # izquierda (fase)

plt.figure(figsize=(6,4))
plt.plot(x, y, label=r"$\cos(x)$")
plt.plot(x, y_down, label=r"$\cos(x) - 1$")
plt.plot(x, y_left, label=r"$\cos(x+\pi/2)$")
plt.title("Transformaciones de la función coseno")
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.ylim(-2,2)
plt.legend()
plt.grid(color='g',linestyle='dotted', linewidth=0.5)
plt.show()


```

# **Transformaciones de Período y Amplitud en Seno y Coseno**

## **Función seno y coseno en forma general**


$f(x) = A \cdot \sin(Bx) \quad \text{o} \quad f(x) = A \cdot \cos(Bx)$


donde:

- \(A\) controla la **amplitud**.
- \(B\) controla el **período**.
$$$$

## **Amplitud**

La **amplitud** es el valor máximo (en valor absoluto) que alcanza la función.


$\text{Amplitud} = |A|$


- Si \(A > 0\), la gráfica conserva su orientación.
- Si \(A < 0\), la gráfica se refleja respecto al eje \(x\).

Ejemplo:

$y = 2\sin(x) \quad \Rightarrow \quad \text{Amplitud } = 2$
$$$$

## **Período**

El **período** es la longitud de un ciclo completo de la función.

$\text{Período} = \frac{2\pi}{B}$

- Si \(B > 1\), la función **se comprime** horizontalmente.
- Si \(0 < B < 1\), la función **se estira** horizontalmente.

Ejemplo:

$y = \sin(2x) \quad \Rightarrow \quad \text{Período } = \pi$
$$$$


$y = \cos\!\left(\tfrac{1}{2}x\right) \quad \Rightarrow \quad \text{Período } = 4\pi$



```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2*np.pi, 2*np.pi, 1000)

# --- Variación de amplitud en seno ---
y1 = np.sin(x)        # amplitud = 1
y2 = 2*np.sin(x)      # amplitud = 2
y3 = -0.5*np.sin(x)   # amplitud = 0.5 y reflexión

plt.figure(figsize=(6,4))
plt.plot(x, y1, label=r"$\sin(x)$")
plt.plot(x, y2, label=r"$2\sin(x)$")
plt.plot(x, y3, label=r"$-0.5\sin(x)$")
plt.title("Efecto de la amplitud en $sin(x)$")
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.legend()
plt.grid(color='g',linestyle='dotted', linewidth=0.5)
plt.show()
```

```python
# --- Variación del período en coseno ---
y1 = np.cos(x)        # período = 2π
y2 = np.cos(2*x)      # período = π
y3 = np.cos(0.5*x)    # período = 4π

plt.figure(figsize=(6,4))
plt.plot(x, y1, label=r"$\cos(x)$")
plt.plot(x, y2, label=r"$\cos(2x)$")
plt.plot(x, y3, label=r"$\cos(0.5x)$")
plt.title("Efecto del período en $cos(x)$")
plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)
plt.legend()
plt.grid(color='g',linestyle='dotted', linewidth=0.5)
plt.show()
```
