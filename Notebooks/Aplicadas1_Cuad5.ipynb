#**Definición**
Dada una función $f(x)$, si $f(x)$ puede hacerse arbitrariamente próximo al número $L$ al tomar $x$ suficientemente cerca de, pero diferente de un número $a$, por la izquierda y por la derecha de $a$, entonces el límite de $f(x)$ cuando $x$ tiende a $a$ es $L$.
$$$$
##Simbolicamente
Si  $~~~~$ $\lim_{x \to a^-}f(x)=L$ $~~~~$ y $~~~~$ $\lim_{x \to a^+}f(x)=L$, $~~~~$ entonces
$~~~~$ $\lim_{x \to a}f(x)=L$
$$$$

#**Definición**

$$$$
Una función $𝑓$ es continua en un punto $𝑐$ si
$$$$

$(i)$ $𝑓$ esta definida en $𝑐$,  $f(c$)
$$$$

$(ii)$ $\lim_{x \to c} f(x)$ existe
$$$$

$(iii)$ $\lim_{x \to c} f(x)=f(c)$
$$$$

Una función $𝑓$ es continua desde la izquierda en un punto $𝑐$ si
$$$$

$$\lim_{x \to c-} f(x)=f(c)$$
$$$$

Una función $𝑓$ es continua por la derecha en un punto $𝑐$ si
$$$$

$$\lim_{x \to c+} f(x)=f(c)$$
$$$$

Una función $𝑓$ es continua en el intervalo abierto $(𝑎, 𝑏)$ si $𝑓$ es
continua en cada punto de $(𝑎, 𝑏)$.
$$$$
Una función $𝑓$ es continua en el intervalo cerrado $[𝑎, 𝑏]$ si $𝑓$ es continua en $(𝑎, 𝑏)$, continua desde la derecha en $𝑎$ y continua desde la izquierda
en $𝑏$.

# **Teorema del valor intermedio**.

Si una función $f$ es continua sobre un intervalo $[a,b]$, $\hspace{0.1cm}$entonces para todo $K$ entre $f(a)$ y $f(b)$,  existe un $c\in[a,b]$ tal que $f(c)=K$.

Por ejemplo para $K=3$ entre $f(0)=1$ y  $f(2)=5$, existe un $c\in[0,2]$ tal que $f(c)=K=3$.

#**Ceros de una función**

Como caso especial, el Teorema del Valor Intermedio nos dice que si $f$ es una función continua en un intervalo cerrado $[𝑎, 𝑏]$ y ademas $f(a)$ y $f(b)$ tienen signos opuestos (es decir, uno es negativo y el otro positivo), entonces hay un punto $c$ en el intervalo abierto $(a,b)$ donde $f(c)=0$.

Ceros.jpg

En estas condiciones, el Teorema del Valor Intermedio
garantiza que la ecuación $f(x)=0$ tiene al menos una solución en $[𝑎, 𝑏]$.

Aunque el teorema no proporciona un método para resolver la ecuación, sí proporciona la base para construir un algoritmo para aproximar una solución con cualquier precisión deseada.

#**Método de la bisección**

Supongamos que 𝑓
es continua en $[𝑎_{1}, 𝑏_{1}]$ $\hspace{0.1cm}$ y $\hspace{0.1cm}$ $𝑓(𝑎_{1})𝑓(𝑏_{1}) < 0$ $\hspace{0.1cm}$ (una forma fácil de comprobar que $𝑓(𝑎_{1}) \hspace{0.1cm} y \hspace{0.1cm} 𝑓(𝑏_{1})$
tienen signos opuestos). Entonces, como se indica arriba, la ecuación
$$$$

$$f(x)=0$$

tiene al menos una solución en $[𝑎_{1}, 𝑏_{1}]$.
$$$$

Sea

$$𝑚_{1}=\frac{𝑎_{1}+𝑏_{1}}{2}$$
$$$$

el punto medio del intervalo.
$$$$

Si $f(𝑚_{1})=0$ significa que hemos encontrado una solución de la ecuación $f(x)=0$.$\hspace{0.1cm}$ Con un error de
$$$$

$$\frac{b_{1}-a_{1}}{2}$$
$$$$

Si $f(𝑚_{1})\neq0$, $\hspace{0.1cm}$ entonces hay dos posibilidades
$$$$

Laprimera es que

$$f(a_{1})f(m_{1})< 0$$
$$$$

En cuyo caso $f(x)=0$ tiene una solución en $[𝑎_{1}, m_{1}]$.
$$$$

LA segunda es que

$$f(m_{1})f(b_{1})< 0$$
$$$$

En cuyo caso $f(x)=0$ tiene una solución en $[m_{1}, b_{1}]$.
$$$$

Bisección.jpg

A continuación se halla el punto medio del intervalo que satisface la condición de signos opuestos, por ejemplo
$$$$

$$𝑚_{2}=\frac{𝑎_{1}+m_{1}}{2}$$
$$$$

Si $f(𝑚_{2})=0$ significa que hemos encontrado una solución de la ecuación $f(x)=0$.$\hspace{0.1cm}$ Con un error de
$$$$

$$\frac{m_{1}-a_{1}}{2}$$
$$$$

Si $f(𝑚_{2})\neq0$, $\hspace{0.1cm}$ entonces de nuevo hay dos posibilidades
$$$$

La primera es que

$$f(a_{1})f(m_{2})< 0$$
$$$$

En cuyo caso $f(x)=0$ tiene una solución en $[𝑎_{1}, m_{1}]$.
$$$$

La segunda es que

$$f(m_{2})f(m_{1})< 0$$
$$$$

En cuyo caso $f(x)=0$ tiene una solución en $[m_{2}, m_{1}]$.
$$$$

y asi se continua iterando hasta alcanzar la precisión deseada.

# **Algoritmo de la bisección**

```python
import numpy as np
import pandas


def bisection(f, a, b, tol):
    # Algoritmo de la bisección para hallar  ceros de una función
    # continua f en un intervalo cerrado [a,b]

    # Verifiquemos que el intervalo [a,b] contenga un cero de f
    if f(a) * f(b) > 0:
        print("El intervalo podría no contener un cero de f, no cambia de signo")
    elif a > b:
      print("El límite izquierdo es mayor al derecho")

    max_iter = 100 # Numero máximo de iteraciones
    # Calcule el punto medio
    m = (a + b)/2
    print("m")

    if np.abs(f(m)) < tol:        # Criterio de parada
        return m
    for i in range(max_iter):
      if f(m) * f(b) < 0:
        a = m #Reemplazo el lado izquierdo del intervalo de prueba
      else: b = m

      m = (a + b)/2
      print(m)

      if np.abs(f(m)) < tol:        # Criterio de parada
        return m


# Función f
def f(x):
  y = x**2-2

  return y

bisection(f,0,2,0.01)
```
