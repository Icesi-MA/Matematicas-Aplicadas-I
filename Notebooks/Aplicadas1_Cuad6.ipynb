#**Introducción**

En esta sección se discute la forma en que se pueden determinar limites de funciones sin recurrir a herramientas de computo. Este desarrollo es importante en el analisis general del comportamiento de funciones y en la construción del concepto de derivada de una función.



#**Teoremas sobre límites**
Dadas $f(x)$   y   $g(x)$ funciones con limites


$$\lim_{x \to c}f(x)=L$$


$$\lim_{x \to c}g(x)=M$$


$g(x)\neq0$, $M\neq0$, $k$ una constante y $p$ un número racional, entonces los siguientes enunciados son verdaderos,


$i)$  $$\lim_{x \to c}k=k$$


$ii)$  $$\lim_{x \to c}x=c$$


$iii)$  $$\lim_{x \to c}kf(x)=k\lim_{x \to c}f(x)=kL$$


$iv)$  $$\lim_{x \to c}[f(x) \pm g(x)]=\lim_{x \to c}f(x)\pm\lim_{x \to c}g(x)=L\pm M$$


$v)$  $$\lim_{x \to c}[f(x)g(x)]=\lim_{x \to c}f(x)\lim_{x \to c}g(x)=LM$$


$vi)$  $$\lim_{x \to c}\frac{f(x)}{g(x)}=\frac{\lim_{x \to c}{f(x)}}{\lim_{x \to c}{g(x)}}=\frac{L}{M}$$


$vii)$  $$\lim_{x \to c}({f(x)})^p=(\lim_{x \to c}{f(x)})^p=L^p$$




# Límites de funciones polinómicas

## Ejemplo 1
Calcular:

$$\lim_{x \to 2} \, (x^2 + 3x - 4)$$

**Solución.** Los polinomios tienen como dominio todo $\mathbb{R}$, así que basta con evaluar en $x=2$:



$$\lim_{x \to 2} (x^2 + 3x - 4) = 2^2 + 3\cdot 2 - 4 = 4 + 6 - 4 = 6$$


$$\lim_{x \to 2} (x^2 + 3x - 4) = 6$$




## Ejemplo 2
Calcular:

$$\lim_{x \to -1} \, (2x^3 + 5x^2 - x + 7)$$

**Solución.**


$$\lim_{x \to -1} (2x^3 + 5x^2 - x + 7)
= 2(-1)^3 + 5(-1)^2 - (-1) + 7
= 2(-1) + 5(1) + 1 + 7$$


$$\lim_{x \to -1} (2x^3 + 5x^2 - x + 7)= -2 + 5 + 1 + 7 = 11$$


$$\lim_{x \to -1} (2x^3 + 5x^2 - x + 7)=11$$

## Ejemplo 3 (límite al infinito)
Calcular:

$$\lim_{x \to \infty} \, (3x^4 - 2x^3 + 7x - 5)$$


**Solución.** En polinomios, el comportamiento al infinito lo determina el **término de mayor grado**. Aquí, $3x^4$ domina:


$$\lim_{x \to \infty} (3x^4 - 2x^3 + 7x - 5) = \infty
\quad \text{(crece como } 3x^4\text{).}
$$

$$\lim_{x \to \infty} (3x^4 - 2x^3 + 7x - 5) = \infty$$



# Límites de funciones racionales

## Ejemplo 1 (límite finito con sustitución directa)
Calcular:

$$\lim_{x \to 2} \frac{x^2 - 4}{x - 2}$$

**Solución.** Sustituimos directamente:


$$\frac{2^2 - 4}{2 - 2} = \frac{0}{0}$$


tenemos una indeterminación. Factorizamos el numerador:

$$x^2 - 4 = (x-2)(x+2)$$


Entonces:

$$\frac{x^2 - 4}{x - 2} = \frac{(x-2)(x+2)}{x-2} = x+2 \quad (x \neq 2)$$


Así:

$$\lim_{x \to 2} \frac{x^2 - 4}{x - 2} = \lim_{x \to 2}(x+2) = 2+2=4$$


$$\lim_{x \to 2}= \frac{x^2 - 4}{x - 2} = 4$$


## Ejemplo 2 (límite al infinito)
Calcular:

$$\lim_{x \to \infty} \frac{3x^3 + 2x}{5x^3 - 7}$$


**Solución.** Dividimos numerador y denominador entre $x^3$(la potencia de $x$ de mayor exponente):


$$\frac{3x^3 + 2x}{5x^3 - 7}
= \frac{3 + \tfrac{2}{x^2}}{5 - \tfrac{7}{x^3}}$$


Cuando $x \to \infty$:


$$\lim_{x \to \infty} \frac{3 + \tfrac{2}{x^2}}{5 - \tfrac{7}{x^3}} = \frac{3}{5}$$


$$\lim_{x \to \infty} \frac{3x^3 + 2x}{5x^3 - 7} = \frac{3}{5}$$


## Ejemplo 3 (límite lateral en asíntota vertical)
Calcular:

$$\lim_{x \to 1^+} \frac{1}{x-1}$$


**Solución.**  


Cuando $x \to 1^+$, el denominador $(x-1) \to 0^+$  

Por lo tanto:

$$\frac{1}{x-1} \to +\infty$$


Así:

$$\lim_{x \to 1^+} \frac{1}{x-1} = +\infty$$



# Límites de funciones con exponentes racionales

## Ejemplo
Calcular:

$$\lim_{x \to 4} \frac{\sqrt{x}-2}{x-4}$$


**Solución.** Multiplicamos numerador y denominadorpor el conjugado:


$$\frac{\sqrt{x}-2}{x-4}\cdot \frac{\sqrt{x}+2}{\sqrt{x}+2}
= \frac{x-4}{(x-4)(\sqrt{x}+2)}
= \frac{1}{\sqrt{x}+2}$$


Entonces,

$$\lim_{x \to 4} \frac{\sqrt{x}-2}{x-4}
=\lim_{x \to 4} \frac{1}{\sqrt{x}+2}=\frac{1}{\sqrt{4}+2} = \frac{1}{4}$$


$$\lim_{x \to 4} \frac{\sqrt{x}-2}{x-4} = \frac{1}{4}$$
