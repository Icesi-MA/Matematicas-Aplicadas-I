**Regla del trapecio**

Para una función $f$ integrable en un intervalo $[a,b]$ tenemos las aproximaciones por la izquierda $I_L$ y por la derecha $I_R$ (Aqui se ha hecho $\Delta x=h$)
$$$$

$$ I_L= h \sum_{i=0}^{n-1} f(c_i)= h \sum_{i=0}^{n-1} f(a+ih) $$
$$$$

$$ I_R= h \sum_{i=0}^{n} f(c_i)=h \sum_{i=1}^{n} f(a+ih) $$
$$$$

con

$$ h=\frac{b-a}{n} $$
$$$$

El promedio de estas aproximaciones es una mejor aproximacion para la integral definida de una función dada y se llama aproximación por regla del trapecio. Esto es
$$$$

$$ \int_{a}^{b} f(x) \,dx= \frac{ I_L+ I_R}{2} $$
$$$$

con un error proporcional a $\frac{1}{n^2}$
$$$$

**Ejemplo**

calcular numericamente la integral

$$ \int_{-1}^{2} (x^2+1) \,dx $$

**Solución**

Utilizando $ n=6 $ tenemos,

$$ h=\frac{2-(-1)}{6}=\frac{1}2{} $$
$$$$

$$ I_R= h \sum_{i=0}^{n} f(c_i)=h \sum_{i=1}^{n} f(a+ih) $$
$$$$

$$ I_R= \frac{1}{2} \sum_{i=1}^{n} f(-1+\frac{1}{2}i) $$
$$$$

$$ I_R= \frac{1}{2} (f(-1+\frac{1}{2}1)+f(-1+\frac{1}{2}2)+f(-1+\frac{1}{2}3)+f(-1+\frac{1}{2}4)+f(-1+\frac{1}{2}5)+f(-1+\frac{1}{2}6)) $$
$$$$

$$ I_R= \frac{1}{2} (f(\frac{-1}{2})+f(0)+f(\frac{1}{2})+f(1)+f(\frac{3}{2})+f(2) )$$
$$$$

$$ I_R= \frac{1}{2} (\frac{5}{4}+1+\frac{5}{4}+2+\frac{13}{4}+5)$$
$$$$

$$ I_R= \frac{1}{2} (\frac{55}{4})$$
$$$$

$$ I_R= \frac{55}{8}= 6.875$$
$$$$

El estudiante debe verificar que (Esta solución debe estar en el laboratorio):
$$$$

$$ I_L=5.375 $$
$$$$

y por consiguiente
$$$$

$$ \int_{a}^{b} f(x) \,dx= \frac{ I_L+ I_R}{2}=\frac{ 5.375+ 6.875}{2}= 6.125 $$
$$$$
