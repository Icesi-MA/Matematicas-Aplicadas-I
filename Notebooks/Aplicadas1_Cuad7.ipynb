#**Introducción**

En el estudio del cálculo surgen dos preguntas fundamentales: ¿Qué tan rápido cambia una función? ¿Cuál es la pendiente de una curva en un punto? y la respuesta a estas preguntas se obtiene mediante el concepto de la derivada. En este cuaderno exploramos la definición de derivada de una función, su interpretación geométrica y el algoritmo para calcular derivadas de funciones mediante la definición de derivada como un límite.
$$$$

# **Definición**

Sea una funcion $f(x)$ definida
$$$$

$$
f:D\subseteq\mathbb{R}\rightarrow\mathbb{R}
$$

La derivada de $f$ en el punto $a$ de su dominio $D$, se define como
$$$$

$$
f'(a)=
\lim_{h\rightarrow0}
\frac{f(a+h)-f(a)}{h},
$$

siempre que dicho límite exista.
$$$$

La derivada de la función $f(x)$ en el punto $a$ expresa la forma en que varía la función en el punto a
$$$$

# **Interpretación Geométrica**

Sea una función $f(x)$ definida
$$$$
$$f:D\subseteq\mathbb{R}\rightarrow\mathbb{R}$$

cuya gráfica esta determinada por los puntos $(x,f(x))$. Sean  $P$ y $Q$ dos puntos de la gráfica de $f$ dados por
$$$$

$$
P(a,f(a))
$$

y

$$
Q(a+h,f(a+h)).
$$
$$$$

La pendiente de la recta secante a la gráfica de $f(x)$ que pasa por $P$ y $Q$ es
$$$$

$$
m_{PQ}=
\frac{f(a+h)-f(a)}{h}.
$$

Cuando

$$
h\rightarrow0,
$$

el punto $Q$ se aproxima al punto $P$ y la recta secante converge hacia la recta tangente a la grafica de $f(x)$ en el punto $a$. Esto es,
$$$$

$$
m_{\text{tangente}}
=
\lim_{h\rightarrow0}
\frac{f(a+h)-f(a)}{h}
$$

$$$$

$$f:D\subseteq\mathbb{R}\rightarrow\mathbb{R}$$
$$$$

$$$$

$$$$

# **Calculo de derivadas utilizando la definición**

A continuación se presentan ejemplos del cálculo de derivadas mediante la definición para cada tipo de funciones.
$$$$

# Funciones constantes

1. $f(x)=10$
$$$$

hallar $f'(2)$
$$$$

Solución

$f'(2)=\lim_{h\to0}\frac{f(2+h)-f(2)}{h}$
$$$$

$f'(2)=\lim_{h\to0}\frac{10-10}{h}$
$$$$

$f'(2)=\lim_{h\to0}\frac{0}{h}$
$$$$

$f'(2)=\lim_{h\to0}0=0$
$$$$

$f'(2)=0$
$$$$

2. $f(x)=-4$
$$$$

hallar $f'(1)$
$$$$

Solución

$f'(1)=\lim_{h\to0}\frac{f(1+h)-f(1)}{h}$
$$$$

$f'(1)=\lim_{h\to0}\frac{-4-(-4)}{h}$
$$$$

$f'(1)=\lim_{h\to0}\frac{0}{h}$
$$$$

$f'(1)=\lim_{h\to0}0=0$
$$$$

$f'(1)=0$
$$$$

# En general

$f(x)=k$, con $k$ una constante arbitraria,
$$$$

$f'(a)=\lim_{h\to0}\frac{f(a+h)-f(a)}{h}$
$$$$

$f'(a)=\lim_{h\to0}\frac{k-k}{h}$
$$$$

$f'(a)=\lim_{h\to0}\frac{0}{h}$
$$$$

$f'(a)=\lim_{h\to0}0=0$
$$$$

$f'(a)=0$, para todo $a$ en el dominio de $f$
$$$$

# Funciones del tipo $f(x)=kx$, con $k$ una constante arbitraria

1. $f(x)=3x$
$$$$

hallar $f'(5)$
$$$$

Solución

$f'(5)=\lim_{h\to0}\frac{f(5+h)-f(5)}{h}$
$$$$

$f'(5)=\lim_{h\to0}\frac{3(5+h)-3(5)}{h}$
$$$$

$f'(5)=\lim_{h\to0}\frac{15+3h-15}{h}$
$$$$

$f'(5)=\lim_{h\to0}\frac{3h}{h}$
$$$$

$f'(5)=\lim_{h\to0}3=3$
$$$$

$f'(5)=3$
$$$$

2. $f(x)=-2x$
$$$$

hallar $f'(3)$
$$$$

Solución

$f'(3)=\lim_{h\to0}\frac{f(3+h)-f(3)}{h}$
$$$$

$f'(3)=\lim_{h\to0}\frac{-2(3+h)+2(3)}{h}$
$$$$

$f'(3)=\lim_{h\to0}\frac{-6-2h+6}{h}$
$$$$

$f'(3)=\lim_{h\to0}\frac{-2h}{h}$
$$$$

$f'(3)=\lim_{h\to0}-2=-2$
$$$$

$f'(3)=-2$
$$$$

# En general

$f(x)=kx$
$$$$

$f'(a)=\lim_{h\to0}\frac{f(3+h)-f(3)}{h}$
$$$$

$f'(a)=\lim_{h\to0}\frac{k(a+h)-k(a)}{h}$
$$$$

$f'(a)=\lim_{h\to0}\frac{ka+kh-ka}{h}$
$$$$

$f'(a)=\lim_{h\to0}\frac{kh}{h}$
$$$$

$f'(a)=\lim_{h\to0}k=k$
$$$$

$f'(a)=k$, para $k$ arbitraria y para todo $a$ en el dominio de $f$
$$$$

# Funciones con potencias

1. $f(x)=x^2$
$$$$

hallar $f'(1)$
$$$$

Solución

$f'(1)=\lim_{h\to0}\frac{f(1+h)-f(1)}{h}$
$$$$

$f'(1)=\lim_{h\to0}\frac{(1+h)^2-1^2}{h}$
$$$$

$f'(1)=\lim_{h\to0}\frac{1^2+2(1)h+h^2-1^2}{h}$
$$$$

$f'(1)=\lim_{h\to0}\frac{2(1)h+h^2}{h}$
$$$$

$f'(1)=\lim_{h\to0}\frac{(2(1)+h)h}{h}$
$$$$

$f'(1)=\lim_{h\to0}{2(1)+h}=2(1)$
$$$$

$f'(1)=2(1)$
$$$$

$f'(1)=2$
$$$$

2. $f(x)=x^3$
$$$$

hallar $f'(1)$
$$$$

Solución

$f'(1)=\lim_{h\to0}\frac{f(1+h)-f(1)}{h}$
$$$$

$f'(1)=\lim_{h\to0}\frac{(1+h)^3-1^3}{h}$
$$$$

$f'(1)=\lim_{h\to0}\frac{1^3+3(1)^2h+3(1)h^2+h^3-1^3}{h}$
$$$$

$f'(1)=\lim_{h\to0}\frac{3(1)^2h+3(1)h^2+h^3}{h}$
$$$$

$f'(1)=\lim_{h\to0}\frac{(3(1)^2+3(1)h+h^2)h}{h}$
$$$$

$f'(1)=\lim_{h\to0}{3(1)^2+3(1)h+h^2}=3(1)^2$

$f'(1)=3(1)^2$
$$$$

$f'(1)=3$
$$$$

# En general

Para todo $a$ en el dominio de $f(x)$.

# **Observación**

En general, la derivada de una función $f(x)$ en un punto $x$ de sudominio, es tambien una funcion $f'(x)$. De cuerdo con esto, se puede deducir la expresion para la derivada de una función en un punto $x$ de su dominio y, a continuación, evaluar el punto particular $x=a$ para obtener la derivada de la función en el punto particular; esto es, $f'(a)$. Veamos como seria en el caso de los ejemplos anteriores.
$$$$

# Para funciones constantes

$f(x)=10$
$$$$

hallar $f'(2)$
$$$$

Solución

$f'(x)=\lim_{h\to0}\frac{f(x+h)-f(x)}{h}$
$$$$

$f'(x)=\lim_{h\to0}\frac{10-10}{h}$
$$$$

$f'(x)=\lim_{h\to0}\frac{0}{h}$
$$$$

$f'(x)=\lim_{h\to0}0=0$
$$$$

$f'(x)=0$
$$$$

Ahora se evalua $x=2$
$$$$

$f'(2)=0$
$$$$

# Para funciones con coeficientes cnstantes

$f(x)=3x$
$$$$

hallar $f'(5)$
$$$$

Solución

$f'(x)=\lim_{h\to0}\frac{f(x+h)-f(x)}{h}$
$$$$

$f'(x)=\lim_{h\to0}\frac{3(x+h)-3(x)}{h}$
$$$$

$f'(5)=\lim_{h\to0}\frac{3x+3h-3x}{h}$
$$$$

$f'(x)=\lim_{h\to0}\frac{3h}{h}$
$$$$

$f'(x)=\lim_{h\to0}3=3$
$$$$

$f'(x)=3$
$$$$

Ahora se evalua $x=5$
$$$$

$f'(5)=3$
$$$$

# Para funciones con potencias

$f(x)=x^2$
$$$$

hallar $f'(1)$
$$$$

Solución

$f'(x)=\lim_{h\to0}\frac{f(x+h)-f(x)}{h}$
$$$$

$f'(x)=\lim_{h\to0}\frac{(x+h)^2-x^2}{h}$
$$$$

$f'(x)=\lim_{h\to0}\frac{x^2+2xh+h^2-x^2}{h}$
$$$$

$f'(x)=\lim_{h\to0}\frac{2xh+h^2}{h}$
$$$$

$f'(x)=\lim_{h\to0}\frac{(2x+h)h}{h}$
$$$$

$f'(x)=\lim_{h\to0}{2x+h}=2x$
$$$$

$f'(x)=2x$
$$$$

Ahora se evalua $x=1$
$$$$

$f'(1)=2(1)=2$
$$$$
