# **Sumas de Riemann**


Supongamos que 𝑓 es continua en [𝑎, 𝑏] y sea $𝑃_𝑛 = {𝑥_0, 𝑥_1, 𝑥_2, . . . , 𝑥_𝑛}$ la partición de [𝑎, 𝑏] usando puntos equidistantes y sean $𝑀_𝑖$ y $𝑚_𝑖$ como se definieron antes. Si elegimos los valores $𝑐_1, 𝑐_2, 𝑐_3, . . . , 𝑐_𝑛$ de modo que
$𝑐_𝑖$ está en el intervalo 𝑖 de la partición (es decir, $𝑥_{𝑖−1} ≤ 𝑐_𝑖 ≤ 𝑥_𝑖)$, se tiene que

$$m_𝑖 ≤ f(𝑐_𝑖) ≤ M_𝑖$$

para $i=1,2,...,n$ y entonces

$$ L(f,P_n)\leq \sum_{i=1}^{n} f(c_i)\Delta x_i \leq U(f,P_n) $$
$$$$

$$ \sum_{i=1}^{n} m_i\Delta x_i \leq \sum_{i=1}^{n} f(c_i)\Delta x_i \leq \sum_{i=1}^{n} M_i\Delta x_i $$
$$$$

y puesto que
$$$$

$$ \sum_{i=1}^{n} m_i\Delta x_i \leq \int_{a}^{b} f(x) \,dx \leq \sum_{i=1}^{n} M_i\Delta x_i $$
$$$$

Si 𝑓 es integrable, siempre es posible elegir una partición 𝑃 tal que
$$$$

$$ | U(f,P_n)-L(f,P_n) | < 𝜖 $$
$$$$

para cualquier 𝜖 > 0 especificado. De ello se deduce que, siempre es posible encontrar, para cualquier 𝜖 > 0 dado, particiones para las cuales

$$ |\int_{a}^{b} f(x) \,dx-\sum_{i=1}^{n} f(c_i)\Delta x_i|< 𝜖 $$

para cualquier elección de los puntos $ 𝑐_1, 𝑐_2, 𝑐_3, . . . , 𝑐_𝑛 $.

La suma

$$\sum_{i=1}^{n} f(c_i)\Delta x_i$$

se llama suma de Riemann y permite obtener aproximaciones para la integral definida con un error $𝜖$, como se vera a cotinuación.
$$$$

**Regla de suma por derecha**

Sea $𝑃_𝑛 = {𝑥_0, 𝑥_1, 𝑥_2, . . . , 𝑥_𝑛}$ la partición de $[a,b]$ usando puntos equidistantes y sea

$$ \Delta x = \Delta x_i =  \frac{b-a}{n}$$

la longitud de los intervalos $[x_{i-1},x_i]$, para $ i=1,2,...,n $.

Observe que

$$\lim_{n\rightarrow\infty}\Delta x =0 $$
$$$$

Si elegimos los puntos $ 𝑐_1, 𝑐_2, 𝑐_3, . . . , 𝑐_𝑛 $ con $ 𝑥_{𝑖−1} ≤ 𝑐_𝑖 ≤ 𝑥_𝑖 $, entonces tenemos, para una función 𝑓 integrable,
$$$$

$$ \int_{a}^{b} f(x) \,dx\approx\sum_{i=1}^{n} f(c_i)\Delta x $$
$$$$

Para hacer esto de manera eficiente es necesario especificar cómo se elegirán los puntos $ 𝑐_1, 𝑐_2, 𝑐_3, ..., 𝑐_𝑛 $. Un método consiste simplemente en elegir $𝑐_𝑖$ como el extremo derecho del intervalo $[𝑥_{𝑖−1}, 𝑥_{𝑖}]$. En ese caso, dado que los puntos en la partición están igualmente espaciados, tenemos
$$$$

$ 𝑐_1 = 𝑥_1 = 𝑥_0 + Δ𝑥 = 𝑎 + Δ𝑥 $

$ 𝑐_2 = 𝑥_2 = 𝑥_1 + Δ𝑥 = 𝑎 + 2Δ𝑥 $

$\vdots$

$ 𝑐_𝑖 = 𝑥_𝑖 = 𝑥_{𝑖−1} + Δ𝑥 = 𝑎 + 𝑖Δ𝑥, $

$ \vdots $

$ 𝑐_𝑛 = 𝑥_𝑛 = 𝑥_{𝑛−1} + Δ𝑥 = 𝑎 + 𝑛Δ𝑥. $
$$$$

Llevando estos puntos a la aproximación de la integral mediante suma de Riemann se tiene
$$$$

$$ \int_{a}^{b} f(x)dx \,\approx \sum_{i=1}^{n} f(c_i)\Delta x $$
$$$$

$$ \int_{a}^{b} f(x) \,dx\approx  \Delta x\sum_{i=1}^{n} f(a+i\Delta x)=I_R $$

donde
$$$$

$$ \Delta x =  \frac{b-a}{n}$$
$$$$

Esta aproximación se conoce como aproximación por la derecha para una integral definida.
$$$$

**Regla suma por izquierda**

El estudiante debe demostrar que se deriva una regla similar utilizando los extremos izquierdos de los intervalos para obtener

$$ \int_{a}^{b} f(x) \,dx \approx\Delta x\sum_{i=0}^{n-1} f(a+i\Delta x)=I_L $$

donde de igual forma

$$ \Delta x =  \frac{b-a}{n}$$

Esta aproximación se conoce como aproximación por la izquierda para una integral definida.
$$$$

**Ejemplo**

calcular numericamente la integral

$$ \int_{-1}^{2} (x^2+1) \,dx $$

**Solución**

Utilizando $ n=6 $ tenemos,

$$ \Delta x=\frac{2-(-1)}{6}=\frac{1}2{} $$
$$$$

$$ I_R=  \sum_{i=1}^{n} f(-1+\frac{1}{2}i)\frac{1}{2} $$
$$$$

Scacando factor comun $\frac{1}{2}$ en la sumatoria

$$ I_R= \frac{1}{2} \sum_{i=1}^{n} f(-1+\frac{1}{2}i) $$
$$$$

$$ I_R= \frac{1}{2} (f(-1+\frac{1}{2}1)+f(-1+\frac{1}{2}2)+f(-1+\frac{1}{2}3)+f(-1+\frac{1}{2}4)+f(-1+\frac{1}{2}5)+f(-1+\frac{1}{2}6)) $$
$$$$

$$ I_R= \frac{1}{2} (f(\frac{-1}{2})+f(0)+f(\frac{1}{2})+f(1)+f(\frac{3}{2})+f(2) )$$
$$$$

$$ I_R= \frac{1}{2} (\frac{5}{4}+1+\frac{5}{4}+2+\frac{13}{4}+5)$$
$$$$

$$ I_R= \frac{1}{2} (\frac{55}{4})$$
$$$$

$$ I_R= \frac{55}{8}= 6.875$$
$$$$

El estudiante debe demostrar que
$$$$

$$ I_L=5.375 $$

A continuación se pesenta el codigo para obtener estas aproximaciones.

```python
import numpy as np

def f(x):

    return (x**2)+1                          # función a integrar

def integral_por_derecha(a, b, n):

  delta_x = (b - a) / n                      # Ancho de cada subintervalo
  suma = 0.0

  for i in range(1, n + 1):                  # Sumas por la derecha
        x_i = a + i * delta_x                # Punto de evaluación
        suma += f(x_i)                       # Evalúa la función en el punto

  integral_approx = delta_x * suma   # Multiplica por el ancho del subintervalo

  return integral_approx

a = -1     # Límite inferior
b = 2      # Límite superior
n = 6      # Número de subintervalos


resultado = integral_por_derecha(a, b, n)
print(f"Aproximación de la integral: {resultado}")
```

```python
import numpy as np

def f(x):

    return (x**2)+1                         # función a integrar

def integral_por_izquierda(a, b, n):

  delta_x = (b - a) / n                      # Ancho de cada subintervalo
  suma = 0.0

  for i in range(n):                         # Sumas por la izquierda
        x_i = a + i * delta_x                # Punto de evaluación
        suma += f(x_i)                       # Evalúa la función en el punto

  integral_approx = delta_x * suma  # Multiplica por el ancho del subintervalo

  return integral_approx

a = -1     # Límite inferior
b = 2       # Límite superior
n = 6     # Número de subintervalos

resultado = integral_por_izquierda(a, b, n)
print(f"Aproximación de la integral: {resultado}")
```
