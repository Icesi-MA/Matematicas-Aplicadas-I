## 1. ¿Qué es un notebook?

Un **notebook** es un documento interactivo compuesto por celdas. En una misma página puede reunir explicaciones, ecuaciones, código, resultados, tablas y gráficas.

- **Jupyter Notebook** es el formato y el entorno abierto en el que se basan estos documentos.
- **Google Colab** es un servicio en la nube que permite abrir y ejecutar notebooks sin instalar Python.
- Este **Jupyter Book** organiza los notebooks como un libro web y permite consultar el material en orden.

### 1.1 Tipos de celda

Una **celda de texto** contiene explicaciones escritas con Markdown y LaTeX. Una **celda de código** contiene instrucciones de Python y muestra su resultado después de ejecutarse.

### 1.2 Editar y ejecutar

En Colab o JupyterLab, haga doble clic sobre una celda de texto para editarla y presione `Shift + Enter` para renderizarla. En una celda de código, `Shift + Enter` ejecuta las instrucciones y avanza a la siguiente celda.

:::{note}
Ejecutar una celda no es lo mismo que guardarla. Guarde periódicamente el archivo `.ipynb` cuando trabaje en Colab o JupyterLab.
:::

## 2. Celdas de texto y Markdown

Markdown es un lenguaje sencillo para estructurar texto sin utilizar un procesador de palabras.

### 2.1 Formato básico

| Sintaxis | Resultado |
| --- | --- |
| `**texto**` | **texto en negrita** |
| `*texto*` | *texto en cursiva* |
| `~~texto~~` | ~~texto tachado~~ |
| `` `código` `` | `código` |
| `[Icesi](https://www.icesi.edu.co/)` | [Icesi](https://www.icesi.edu.co/) |

### 2.2 Encabezados

Los símbolos `#` indican niveles de organización. El título principal usa un solo `#`; las secciones usan `##` y las subsecciones usan `###`.

```markdown
# Título principal
## Sección
### Subsección
```

Use los niveles en orden y evite saltar, por ejemplo, de `##` a `####`.

### 2.3 Listas

```markdown
- Primer elemento
- Segundo elemento
  - Elemento anidado

1. Primer paso
2. Segundo paso
```

Resultado:

- Primer elemento
- Segundo elemento
  - Elemento anidado

1. Primer paso
2. Segundo paso

### 2.4 Tablas

```markdown
| Clase | Día | Tema |
| --- | --- | --- |
| 1 | Lunes | Markdown |
| 2 | Miércoles | LaTeX |
```

| Clase | Día | Tema |
| --- | --- | --- |
| 1 | Lunes | Markdown |
| 2 | Miércoles | LaTeX |

## 3. Escritura matemática con LaTeX

LaTeX permite escribir matemáticas con una notación clara y consistente.

### 3.1 Expresiones dentro de una oración

Use un signo `$` a cada lado. Por ejemplo, la entrada `$f(x)=x^2+1$` produce $f(x)=x^2+1$.

### 3.2 Ecuaciones centradas

Use dos signos `$$` a cada lado:

```latex
$$
\lim_{x \to 2}\frac{x^2-4}{x-2}=4
$$
```

El resultado es:

$$
\lim_{x \to 2}\frac{x^2-4}{x-2}=4
$$

### 3.3 Símbolos frecuentes

| Entrada | Resultado |
| --- | :---: |
| `\le` | $\le$ |
| `\approx` | $\approx$ |
| `\in` | $\in$ |
| `\sqrt{x}` | $\sqrt{x}$ |
| `\frac{x}{y}` | $\frac{x}{y}$ |
| `\sum_{i=1}^{n} i` | $\sum_{i=1}^{n} i$ |
| `\int_a^b f(x)\,dx` | $\int_a^b f(x)\,dx$ |

Para consultar más ejemplos, revise la [guía de Markdown](https://www.markdownguide.org/basic-syntax/) y la [referencia de símbolos matemáticos](https://en.wikibooks.org/wiki/LaTeX/Mathematics).
