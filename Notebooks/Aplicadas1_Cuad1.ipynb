* **Jupyter Notebook** es una interfaz web de código abierto que permite la inclusión de texto en Markdown, enunciados formales en LaTeX, vídeo, audio, imágenes, así como la ejecución de código a través del navegador y en múltiples lenguajes, entre ellos **Python**.

* Los documentos Jupyter Notebook constan de celdas de texto y celdas de código (Python), convirtiendo este documento en un poderoso entorno interactivo, donde tenemos en un sólo documento, código, ecuaciones, gráficas, discusiones y comentarios

* **Google Colaboratory** o **Colab** Es un producto de Google Reseach alojado en Juppiter Notebook que ofrece recursos informáticos y permite al usuario escribir y ejecutar código de Python en el navegador, brindando integración directa con servicios como google drive, GitHub, acceso a hardware potente para computación de alto rendimiento, etc.
