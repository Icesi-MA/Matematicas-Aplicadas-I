## 2. Variables y tipos de datos

Una variable asocia un nombre con un valor. Use nombres descriptivos, sin espacios y sin comenzar con un número.

Python incluye, entre otros, estos tipos básicos:

- `int`: números enteros;
- `float`: números con parte decimal;
- `str`: texto;
- `bool`: valores lógicos `True` o `False`.

La función `type` permite consultar el tipo de un valor.

````python
cantidad = 5
temperatura = 24.5
nombre = "Ana"
curso_activo = True

print(type(cantidad))
print(type(temperatura))
print(type(nombre))
print(type(curso_activo))
````

### 2.1 Asignación y reasignación

El símbolo `=` asigna un valor. Si se vuelve a utilizar con el mismo nombre, el valor anterior se reemplaza.

````python
a = 5
print(a)

a = 1974
print(a)
````

### 2.2 Conversión de tipos

Las conversiones deben solicitarse explícitamente. `int(5.23)` produce `5` porque elimina la parte decimal; no realiza un redondeo.

````python
decimal = 5.23
entero = int(decimal)

print(decimal)
print(entero)
````

## 3. Operaciones aritméticas

Python puede utilizarse como una calculadora. Observe que la potencia se escribe con `**`; el símbolo `^` tiene otro significado en Python.

````python
print(3 + 24)   # Suma
print(5 - 14)   # Resta
print(3.5 * 2)  # Multiplicación
print(2 / 3)    # División
print(2**3)     # Potencia
````

### 3.1 Prioridad de las operaciones

Python evalúa primero paréntesis, después potencias, luego multiplicaciones y divisiones, y finalmente sumas y restas.

Antes de ejecutar la siguiente celda, intente predecir cada resultado.

````python
print(2 * 3 + 1)
print(2 * (3 + 1))
print(2**3 * 2)
````

### 3.2 Potencias y raíces

Una raíz puede escribirse como una potencia racional. La raíz cuadrada de un número negativo no es real; Python puede representarla como un número complejo.

````python
print(9**(1/2))
print((-5)**(1/2))
````

## 4. Funciones y módulos matemáticos

Un módulo reúne funciones relacionadas. En este curso utilizaremos principalmente:

- `math`, para cálculos escalares;
- `numpy`, para trabajar eficientemente con muchos valores.

Las funciones trigonométricas reciben ángulos en **radianes**.

````python
import math
import numpy as np

print(math.sqrt(5))
print(math.sin(math.pi / 2))
print(np.exp(1))
````

## 5. Listas en Python

Una lista almacena una colección ordenada de elementos. Puede contener valores de distintos tipos y puede modificarse.

````python
datos = [10, 20, 30, 40]
print(datos)
````

### 5.1 Índices

Python comienza a contar desde cero.

```text
Valores:   10  20  30  40
Índices:    0   1   2   3
Negativos: -4  -3  -2  -1
```

````python
print(datos[0])   # Primer elemento
print(datos[1])   # Segundo elemento
print(datos[-1])  # Último elemento
````

### 5.2 Segmentos de una lista

La sintaxis `inicio:fin` incluye el índice inicial y excluye el final.

````python
print(datos[1:3])
print(datos[:2])
print(datos[2:])
print(datos[:])
````

## 6. Secuencias y ciclos

La clase `range` genera una secuencia de enteros. El valor final no se incluye.

Esta sección es una primera aproximación a los ciclos; no es necesario memorizar toda la sintaxis en la primera lectura.

````python
print(list(range(5)))
print(list(range(2, 8)))
print(list(range(0, 10, 3)))
````

### 6.1 Ciclos `for`

Un ciclo `for` repite un bloque de instrucciones para cada elemento de una secuencia. La sangría indica qué instrucciones pertenecen al ciclo.

````python
for i in range(5):
    print(i, i**2)
````

## 7. Arreglos de NumPy

Una lista y un arreglo de NumPy no se comportan igual. Multiplicar una lista repite sus elementos; multiplicar un arreglo aplica la operación a cada valor.

````python
import numpy as np

lista = [1, 2, 3]
arreglo = np.array([1, 2, 3])

print(lista * 2)
print(arreglo * 2)
````

### 7.1 Operaciones vectorizadas

NumPy puede evaluar una función sobre todos los elementos de un arreglo sin escribir un ciclo explícito.

````python
valores = np.array([0, np.pi / 2, np.pi])
print(np.sin(valores))
print(np.square([2, 3, 6]))
````

### 7.2 Puntos igualmente espaciados con `linspace`

`np.linspace(inicio, fin, cantidad)` genera una cantidad determinada de puntos entre dos extremos, incluyendo ambos.

````python
x = np.linspace(1, 2, 5)
print(x)
````

## 8. Primera gráfica con Matplotlib

Matplotlib es una biblioteca de visualización. Utilizaremos el módulo `pyplot`, normalmente importado con el alias `plt`.

Para representar una función continua, generamos una muestra finita de valores de $x$, evaluamos la función y unimos los puntos.

````python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2, 2, 100)
y = x**2

plt.plot(x, y, label="y = x²")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Gráfica de y = x²")
plt.grid()
plt.legend()
plt.show()
````
