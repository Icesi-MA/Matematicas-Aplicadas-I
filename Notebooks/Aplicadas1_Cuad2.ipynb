# Listas en Python
Las Listas son estructuras de datos incorporadas en Python que actúan como contenedores para el almacenamiento de una colección de elementos que son accesibles mediante indexación.

Una lista en Python es:

1. **Ordenada:** esto quiere decir que los elementos dentro de ella están indexados y se accede a ellos a través de una locación indexada.
2. **Editable:** los elementos dentro de una lista pueden editarse, añadir nuevos o eliminar los que ya tiene.  
3. **Dinámica:** las listas pueden contener diferentes tipos de datos y hasta de objetos.

Podemos declarar una lista vacía de las siguentes dos maneras:

````python
lista_1 = []              # Opción 1
lista_2 = list()          # Opción 2
print(lista_1)
print(lista_2)
````

Para acceder a un elemento de una lista se utiliza su índice. El índice es un número entero que indica la posición de un elemento en una lista. El primer elemento de una lista siempre comienza en el índice 0.   Por ejemplo, en una lista con 4 elementos, los índices de cada uno de los ítems serían 0, 1, 2 y 3.

````python
Lista1=[1,2,3,4,5,6]
Lista1[0]                 # Primer elemento de la lista
````

````python
Lista1[1]                 # Segundo elemento de la lista
````

En Python está permitido usar índices negativos para acceder a los elementos de una secuencia. En este caso, el índice -1 hace referencia al último elemento de la secuencia, el -2 al penúltimo y así, sucesivamente

````python
Lista1[-1]               # Ultimo elemento de la lista
````

````python
Lista1[-2]               # Penultimo elemento de la lista
````

También es posible acceder a un subconjunto de elementos de una lista utilizando rangos en los índices. Esto es usando el operador `[:]`

````python
Lista1[2:5]              # Elementos desde el índice 2 hasta el índice 5 (no incluidos)
````

````python
Lista1[:]                # Todos los elementos
````

````python
Lista1[1:]                # Elementos desde el índice 1 (no incluido)
````

````python
Lista1[:3]                # Elementos hasta el índice 3 (no incluido)
````

# La clase `range`  
`range` en Python es un tipo que se utiliza para representar una secuencia inmutable de números. Uno de sus principales usos es junto a la sentencia `for`, para definir un bucle sobre el que se itera un número determinado de veces.  

Para crear un objeto de tipo range, se pueden usar dos constructores :

`range(fin)`: Crea una secuencia numérica que va desde 0 hasta fin - 1.

`range(inicio, fin, [paso])`: Crea una secuencia numérica que va desde inicio hasta fin - 1 y la secuencia genera los números de paso en paso.  
$$$$
Veamos

````python
list(range(10))          # se genera una lista desde 0 hasta 9
````

````python
list(range(5, 10))       # segenera una lista desde 5 hasta 9
````

````python
list(range(0, 10, 3))    # se geenra una lista desde 0 hasta 9 de 3 en 3
````

En los 3 ejemplos anteriores se ha usado la clase `range` como argumento de `list`, esto se hace para que se muestre, en pantalla, la secuencia completa que se genera con range.

Los ciclos `for` pueden combinarse con la función `range` de Python, para generar un arreglo secuencial en lugar de escribir el contenido de nuestro arreglo término a término:
$$$$
Veamos

````python
# Ejemplo1: ciclo for declarando todos los elemntos del arreglo
for i in [0,1, 2, 3, 4]:
    print(i,end=", " )         # imprime los valores: 0,1, 2, 3, 4,
````

````python
# Ejemplo2: ciclo for combiado con la función range.
for i in range(5):             # generamos el arreglo usando la función range.
    print(i,end=",")           # imprime los valores: 0, 1, 2, 3, 4,
````

````python
# Ejemplo3 ciclo for aplicando una instrucción
for i in [1, 3, 5, 7, 9]:
    x = i**2 - (i-1)*(i+1)
    print(x, end=", ")          # imprime 1, 1, 1, 1, 1,
````

````python
# Ejemplo4 Ciclo for mediante la función rangue y aplicando una instrucción
for i in range(1,10,):
  x = i**2 - (i-1)*(i+1)
  print(x, end=", ")             # imprime 1, 1, 1, 1, 1,
````

# Módulo Matplotlib

Como es natural de pensar no podemos generar en una computadora conjuntos continuos (pues necesitariamos cantidad infinita de memoria), por tanto, si queremos generar gráficas de funciones cuyo dominio es un conjunto continuo, debemos tomar una muestra (un subconjunto finito lo suficientemente "bueno") y aplicar la función a los elementos de esa muestra y posteriormente, con ayuda de la represenación visual, emulamos la aplicación de la función a un conjunto infinito de valores.  

**Matplotlib** es una librería de visualización de python creada por Jhon Hunter la cual está especialmente diseñada para visualizar datos. Tomando inspiración de MATLAB matplotlib trabaja con tres elementos principales: axis (“ejes”), figure (“figura”) y plot (“gráfico” o “cuadro”).

La libreria **Matplotlib** tiene una gran cantidad de módulos, pero nosotros estaremos interesados en usar únicamente el submódulo **pyplot** y para instalarlo la primera línea de código debería ser la siguiente:

```
import matplotlib.pyplot as plt
```

Recuerda que al escribir

```
import matplotlib
```

estariamos importando toda la libreria, pero el agregar  **.pyplot** nos asegura que sólo vamos a importar el módulo que queremos.

La primera función que vamos a utilizar es la función plot, esta función sirve para graficar líneas continuas y es de las más sencillas de utilizar. La función plot sigue la siguiente sintaxis

```
plt.plot(ARGUMENTOS) # Los argumentos se separan por comas
```

Ahora deberiamos definir dos arrays, cada uno conteniendo un conjunto  de números. Estos arrays los pasaremos como argumento a la función plot y ella nos devolverá un gráfica con una línea la cual une todos los puntos, tomando los elementos del conjunto como las coordenadas x e y de los distintos puntos.
