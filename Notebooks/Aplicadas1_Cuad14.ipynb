# **Introducción**

La diferenciación y la integración son inversas. Si una función $F$ se define como la integral de $f$ desde $a$ hasta $x$, entonces la derivada de $F$ es $f$.


$$F(x)=\int_a^x f(x)\,dx.$$

siempre que

$$f(x)=F'(x)$$

Esta relación se establece formalmente mediante los siguientes resultados fundamentales del cáculo integral.

# Teoremas Fundamentales del Cálculo
1.   Sea $f$ cualquier función continua en $[a, b]$. Entonces $f$ es la derivada de alguna función diferenciable. De hecho, para $x$ en $[a, b]$,

$$ \frac{d}{dx}\left(\int_a^x f (t)\,dt\right)=f(x).$$

2.   Sea $F$ cualquier función con una derivada continua en $[a, b]$. Luego $$ F(b)-F(a)=\int_a^b F' (x)\,dx.$$



# Observación

Es necesario tener en cuenta que métodos para el cálculo númerico de integrales como el de: punto medio, trapecio, Simpson, se basan en la idea de hacer particiones de un intervalo, formar rectangulos cuya base sea la longitud de la partición, calcular las áreas de cada rectangulo y realizar la suma, para de esta forma encontrar una aproximación de la integral. El siguiente bosquejo de la demostración del segundo teorema fundamental nos ayuda a entender en que se basa cada uno de los métodos mencionados anteriormente.



Considere una partición arbitraria del intervalo $[a,b]$, esto es,  $a=x_0 < x_1 < x_2 < · · · < x_{n−1} < b=x_n$. Supongamos que $ \Delta x_i= x_i-x_{i-1},\,i=1,2,...,n.$ Ahora,

$$ F(b)-F(a) = F(x_n)-F(x_0)$$

$$ F(b)-F(a) = F(x_n)+(F(x_{n-1})-F(x_{n-1}))+...+(F(x_{1})-F(x_1))-F(x_0)$$

$$ F(b)-F(a)=(F(x_n)-F(x_{n-1}))+(F(x_{n-1})-F((x_{n-2}))+...+(F(x_{1})-F(x_0))$$

$$ F(b)-F(a)=\sum_{i=1}^n F(x_i)-F(x_{i-1}).$$

Por el Teorema del Valor Medio para derivadas existe un $c_i$ en cada intervalo $[x_i,x_{i-1}]$ tal que

$$F'(c_i)=\frac{F(x_i)-F(x_{i-1})}{x_i-x_{i-1}}.$$

Por el primer teorema fundamental del cálculo se obtiene que

$$F(x_i)-F(x_{i-1})=f(c_i)(x_i- x_{i-1}) $$

Todo lo anterior nos conduce a la aproximación de la integral,

$$ F(b)-F(a)=\sum_{i=1}^n  f(c_i)\Delta x_i$$


# **Antiderivadas y su relación con la derivada**

En Cálculo Diferencial aprendimos que si conocemos una función
$$
F(x)
$$
podemos calcular su derivada
$$
F'(x).
$$
Ahora nos planteamos el problema inverso. Si conocemos
$$
f(x),
$$
¿podemos encontrar una función cuya derivada sea precisamente $f(x)$?
Por ejemplo, sabemos que
$$
\frac{d}{dx}(x^3)=3x^2.
$$
Entonces, dada la función
$$
3x^2,
$$
¿cómo recuperar la función original? Esta pregunta da origen al concepto de **antiderivada**.

# Definición
Sea $f$ una función definida sobre un intervalo $I$. Una función $F$ se llama **antiderivada** de $f$ sobre $I$ si
$$
F'(x)=f(x)
$$
para todo
$$
x\in I.
$$
En otras palabras, derivar $F$ produce exactamente $f$.

# Notación
La familia completa de antiderivadas se escribe como
$$
\int f(x),dx.
$$
No representa todavía un área. Representa **el conjunto de todas las funciones cuya derivada es $f(x)$.**

Considerese
$$
F(x)=x^2.
$$
Su derivada es
$$
2x.
$$
Pero también
$$
G(x)=x^2+5
$$
tiene derivada
$$
2x.
$$
Igualmente,
$$
H(x)=x^2-13
$$
también posee la misma derivada. En realidad,
todas las funciones
$$
x^2+C
$$
tienen derivada
$$
2x.
$$
Por ello,
$$
\int 2x,dx=x^2+C.
$$
La constante (C) recibe el nombre de **constante de integración**.

A continuación se presenta graficamente algunas antiderivadas de $2x$. Todas las curvas tienen exactamente la misma pendiente en cada punto. Lo único que cambia es su desplazamiento vertical determinado or el valor de la constante $C$.

```python
import numpy as np
import matplotlib.pyplot as plt

x=np.linspace(-3,3,400)

plt.figure(figsize=(8,5))

for C in [-4,-2,0,2,4]:
    plt.plot(x,x**2+C,label=f"C={C}")

plt.title("Familia de antiderivadas de 2x")
plt.xlabel("x")
plt.ylabel("F(x)")
plt.grid(True)
plt.legend()
plt.show()
```

# **Aplicaciones de la integral**
Una ecuación de la forma

$$ \frac{dx}{dt}=\dot{x}(t)=f(t,x(t))\qquad (1)$$

que depende de una única variable independiente, $t$, se denomina **ecuación diferencial ordinaria (EDO)**.

Resolver una ecuación de este tipo implica identificar una función $𝑥$, definida únicamente en términos de la variable independiente $𝑡$, que satisfaga la relación de $(1)$ para todo $𝑡$ en algún intervalo abierto. Para muchas ecuaciones, no es posible encontrar una solución exacta y entonces se buscan buenas aproximaciones.

Todos os problemas en física e ingeniería se modelan mediante ecuaciones diferenciales. La solución de una EDO depende del tipo particular de ecuación diferencial.

**Por ejemplo**
La función de velocidad en caída libre viene dada por la ecuación diferencial

$$\dot{x}(t)=v_0+g t$$

donde $g$ es la constante de gravitación universal. Es fácil ver entonces que esta es una ecuación diferencial cuya solución se obtiene al integrar la función $\dot{x}(t)$.

El caso mas simple podría ocurrir cuando en la ecuación $(1)$ tenemos que $f(x(t),t)=g(t)$, puesto que la podemos resolver integrando. En efecto,

$$ \int_{t_0}^{t}\dot{x}(s)\,ds=\int_{t_0}^{t} g(s)\,ds\qquad (1)$$
o equivalentemente
$$x(t)=x(t_0)+\int_{t_0}^{t} g(s)\,ds.$$

# Variables separables


Es un técnica para encontrar soluciones en forma cerrada para  EDOs,
es decir, soluciones expresables en términos de las funciones elementales del cálculo. Consideraremos ecuaciones de la forma#

$$ \frac{dx}{dt}=f(t,x(t))=g(x)h(t).\qquad (2)$$

Note que podemos reescribr la ecuación como

$$ \frac{\dot x}{g(x)}=h(t)$$

en todos los puntos donde $g(x)\not=0$. Ahora bien, integrando entre $t_0$ y $t$ obtenemos

$$ \int_{t_0}^t\frac{1}{g(x(s))}\dot x(s)\, ds=\int_{t_0}^th(s)\,ds$$

Utilizando el cambio de variable $u=x(s)$ tenemos

$$ \int_{t_0}^t\frac{1}{g(x(s))}\dot x(s)\, ds=\int_{x_0}^x\frac{1}{g(u)}\, du=\int_{t_0}^th(s)\,ds.\quad (3)$$

Por lo tanto, podemos resolver una ecuación con variables separables siempre que podamos evalúar las integrales en $(3)$ y luego resolver la ecuación para $𝑥$.

# Tecnicas numéricas

La imposibilidad de resolver una ecuación diferencial se debe en parte a la imposibilidad de integrar la mayoría de las funciones. Si se tuviera que elegir funciones para integrar al azar, encontraríamos que la gran mayoría no tiene antiderivadas. La historia en las EDO es la misma: elija cualquier combinación de una función, sus derivadas y otras funciones de forzamiento y encontrará que no hay forma de llegar a una solución analítica que involucre las operaciones y funciones regulares de las matemáticas: combinaciones lineales, potencias , raíces, funciones trigonométricas, logaritmos, etc. Hay teoremas de ecuaciones diferenciales que garantizarán la existencia y unicidad de soluciones a muchas ecuaciones diferenciales, pero el simple hecho de saber que la solución existe no es suficiente para descubrirla. Las técnicas numéricas nos brindan una vía para al menos aproximarnos a estas soluciones.

# Método de Euler
La base de la mayoría de métodos numéricos para encontrar soluciones aproximadas de una ecuación diferencial
$$
dxdt=f(x,t),x(0)=x0
$$
es reemplazar la derivada por un cociente de diferencias, esto es,
$$
x(t+h)−x(t)h≈f(x,t),x(0)=x0
$$
lo cual se reescribe como
$$
x(t+h)−x(t)≈hf(x,t)
$$
x(t+h)≈x(t)+hf(x,t)
$$
Este es el denominado método de Euler para soluciones aproximadas.

```python
import numpy as np
import math
import matplotlib.pyplot as plt
import pandas as pd

real_x = []
t_real = np.linspace(0,5,num=1000)

def euler(f,x0,t0,tmax,n):
    t = np.linspace(t0, tmax, n)# defina el dom up the domain based on t0, tmax, and n
    h = (tmax-t0)/n
    # next set up an array for x that is the same size a t
    x = np.zeros_like(t)
    x[0] = x0# fill in the initial condition
    for i in range( n-1 ): # think about how far we should loop
        x[i+1] = x[i]+h*f(x[i],t[i]) # advance the solution forward in time with Euler
    return x, t

edo = lambda x,t: (-0.71)*(x-70)
f = lambda t: 28.6*np.exp(-0.71*t)+70

for i in t_real:
  real_x.append(f(i))
```

```python
errors = []

x, t = euler(edo, 98.6, 0,5, 50)

for i in range(5):
  errors.append((f(t[i]) - x[i]) / f(t[i]))

plt.plot(t_real,real_x, 'r.-', label='real')
plt.plot(t,x,'b.-', label='aprox')
plt.rcParams["figure.figsize"] = (12,7)
plt.legend()
plt.xlabel("t")
plt.ylabel("x(t)")

plt.grid()
plt.show()
```
