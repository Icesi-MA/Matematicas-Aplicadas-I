# **1. Introducción**

Ya hemos explorado la derivada no solo como una regla de cálculo, sino como la **mejor aproximación afín** de una función en la vecindad de un punto. Esta perspectiva local es crucial para entender cómo predecir el comportamiento global de un sistema y cómo identificar puntos críticos en optimización.


En general la optimización consiste en encontrar donde se dan los valores extremos de una función; esto es, los valores máximos y mínimos de una función dada y además determinar cuales son esos valores máximos y mínimos, si es que existen. Sin embargo, en la práctidca determinar esos valores extremos esta sujeto a ciertas restricciones particulares de cada contexto o problema considerado.

En este laboratorio veremos la forma en que la mejor aproximación afin de una función nos permite plantear un modelo para optimización de funciones.


# **2. Aproximación Afín**

La derivada de una función $f(x)$ en un punto $c$ permite obtener la mejor aproximación afín de la función $f(x)$ en la vecindad del punto $c$:


$$T(x) \approx f(c) + f'(c)(x - c)$$


Esta aproximación afin $T(x)$ permite aproximar valores locales de la función $f(x)$, identificar crecimiento y decrecimiento local de la  $f(x)$ y por consiguiente formular los problemas de optimización.


**Ejemplo**

Considere la función


$$f(x)=x^2-4x+5$$


Determine la mejor aproximación afín a $f(x)$ en $c=1$, en $c=3$ y en $c=2$.


A continuación corra el siguiente codigo,

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return x**2 - 4*x + 5

def df(x):
    return 2*x - 4

c = 1 # punto de aproximación

x = np.linspace(-1, 5, 100)
y = f(x)

# recta tangente
tangent = f(c) + df(c)*(x - c)

plt.figure()
plt.plot(x, y, label='f(x)')
plt.plot(x, tangent, '--', label='Aproximación afín')
plt.scatter(c, f(c))
plt.legend()
plt.title("Aproximación afín")
plt.grid()
plt.show()
```

Ahora modifique el valor de $c$ y corra el codigo para $c=3$ y para $c=2$.


Observe que en cada caso la aproximación afín determina si la función $f(x)$ es dereciente, creciente, o ni lo uno ni lo otro. El punto en el que la aproximación afin no determina si la función crece o decrece se denomina punto crítico y en este caso, dicho punto crítico es un mínimo de la función en el intervalo considerado en la gráfica.




```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return x**2 - 4*x + 5

def df(x):
    return 2*x - 4

c = 2 # punto de aproximación

x = np.linspace(-1, 5, 100)
y = f(x)

# recta tangente
tangent = f(c) + df(c)*(x - c)

plt.figure()
plt.plot(x, y, label='f(x)')
plt.plot(x, tangent, '--', label='Aproximación afín')
plt.scatter(c, f(c))
plt.legend()
plt.title("Aproximación afín con c = 2")
plt.grid()
plt.show()
```

# **3. Valores extremos de una función**

$i)$ Un número $f(c)$ es un máximo absoluto de una función $f$ si para todo $x$ en el dominio de $f$,   $f(x) \leq f(c)$.

$ii)$ Un número $f(c)$ es un mínimo absoluto de una función $f$ si para todo $x$ en el dominio de $f$,   $f(x) \geq f(c)$.

# Definición

Si $𝑓$ es diferenciable en $𝑐$ y $𝑓^{\prime}(𝑐) = 0$, entonces llamamos a $𝑐$ un punto crítico o punto estacionario de $𝑓$. Un punto $𝑐$ en el que la derivada de $𝑓$ no está definida se llama punto singular de $𝑓$.


De acuerdo con esto, los candidatos para la ubicación de los valores extremos de una función continua en un intervalo cerrado se dividen en tres categorías: $(a)$ puntos limite del intervalo, $(b)$ puntos críticos y $(c)$ puntos singulares. Para determinar los valores extremos de una función $f$ identificamos estos puntos, evaluamos la funcion en dichos puntos y determinamos el mayor y el menor valor de la función.

# Ejemplo

Encontrar los valores máximo y mínimo absolutos de la función


$$g(x)=x^3-3x$$


En el intervalo $[-1.5,1.5]$



**Solucion**

Hallamos los puntos críticos


$$g'(x)=3x^2-3$$

$$0=3x^2-3$$

$$3=3x^2$$

$$\sqrt{1}=x$$


En el intervalo considerado esto corresponde a $x=\pm 1$.

Puesto que no hay puntos singulares, evaluamos $g(x)$ en los puntos límite del intervalo y en los puntos críticos.


$$g(-1.5)=(-1.5)^3-3(-1.5)=1.125$$


$$g(-1)=(-1)^3-3(-1)=2$$


$$g(1)=(1)^3-3(1)=-2$$


$$g(1.5)=(1.5)^3-3(1.5)=-1.125$$


y se terminan los valores extremos de $g(x)$ y su ubicación en el intervalo considerado.


Hay un maximo $g(x)=2$en $x=-1$ y un mínimo $g(x)=-2$ en $x=1$. Estos resultados pueden verse en la siguiente gráfica.

# **5. Extremos locales**

Si estudiamos el comportamiento de funciones en intervalos abiertos no podemos hablar de valores extremos absolutos, debemos hablar de valores extremos locales, como se vera en la siguientes proposiciones.


# Proposición 1

Supongamos que $𝑓$ y $𝑓^{\prime}$ son continuas en un intervalo abierto (𝑎, 𝑏) y 𝑐 es el único punto crítico de 𝑓 en (𝑎, 𝑏). Si  𝑓(𝑐) es el valor mínimo de 𝑓 en (𝑎, 𝑏) entonces 𝑓 tiene un mínimo local en 𝑐; si 𝑓(c) es el valor máximo de 𝑓 en (𝑎, 𝑏) entonces 𝑓  tiene un máximo local en 𝑐.
​

# Proposición 2

Supongamos que 𝑓, 𝑓^{\prime} y 𝑓^{\prime\prime} son continuas en un intervalo abierto (𝑎, 𝑏) y que 𝑐 es un punto crítico de 𝑓 en (𝑎, 𝑏). Entonces 𝑓 tiene un máximo local en 𝑐 si 𝑓^{\prime\prime}(𝑐) < 0 y 𝑓 tiene un mínimo local en 𝑐 si 𝑓^{\prime\prime}(𝑐) > 0.




Para trabajar con el criterio de la segunda derivada vamos a introducir la biblioteca SymPy de Python para matemáticas simbólicas.


# Ejemplo

Hallar los puntos críticos de la funcion dada y determinar si son maximo o mínimo locales.


$$f(x)=5x^3-3x^5 $$


# Solucion

Primero importamos los modulos necesarios

```python
from sympy import *                # Importa todas las funciones y clases de la biblioteca SymPy
from sympy.plotting import plot    # Importa la función plot de la subbiblioteca plotting de SymPy
from sympy.abc import x            # Importa el símbolo x de la subbiblioteca abc de SymPy
init_printing()                    # Inicializa la impresión de salida para las expresiones matemáticas
```

```python
f = 5*x**3-3*x**5
```

La función "diff" calcula la primera y segunda derivada

```python
df = diff(f, x)                    # 1era. derivada
d2f = diff(f,x,2)                  # 2nda derivada
df, d2f                            # Imprime
```

Se calculan los puntos críticos resolviendo $f'(x)=0$

```python
pc = solve(Eq(df, 0))             # Resuelve la ecuación df=0
pc
```

Se determina si se trata de un minimo o un maximo mediante el criterio de la segunda derivada



```python
d2f.subs(x,pc[0])                # d2f(-1)=?
```

$f''(-1)=30(-1)(1−2(-1)^2))=30>0$,    $x=-1$ es un mínimo local

```python
f.subs(x,pc[0])                 # d2f(-1)=?
```

y su valor es $f(-1)=5(-1)^3-3(-1)^5=-2$

```python
d2f.subs(x,pc[1])               #  # d2f(0)=?
```

El criterio no brinda información sobre la naturaleza de este punto crítico

```python
d2f.subs(x,pc[2])              # d2f(1)=?
```

$f''(1)=30(1)(1−2(1)^2))= -30< 0$,    $x=1$ es un máximo local

```python
f.subs(x,pc[2])
```

y su valor es $f(1)=5(1)^3-3(1)^5=2$

Por lo tanto, 𝑓 tiene un mínimo local de valor −2 en 𝑥 = −1 y un máximo local de valor 2 en 𝑥 = 1. Desafortunadamente, 𝑓^{\prime\prime}(0) = 0, por lo que la prueba de la segunda derivada no nos da información sobre la naturaleza del punto 0. # **6. Optimización**

Vamos a describir el proceso de optimización mediante un ejemplo.


# Problema

Una empresa desea **maximizar su beneficio** al producir cierta cantidad de unidades $(x)$.

La función de beneficio está dada por:



$$f(x) = -x^2 + 6x$$


Sin embargo, existe una restricción de producción:


$$0 \leq x \leq 5$$


# Solución Analítica

Derivada de la función


$$f'(x) = -2x + 6$$

Para encontrar puntos críticos igualamos a cero:


$$-2x + 6 = 0$$


$$x = 3$$

Verificar si está en la restricción

El valor $(x=3)$ pertenece al intervalo $[0, 5]$, por lo tanto es candidato válido.

Evaluar en extremos y punto crítico


$$f(0) = 0$$


$$f(3) = -9 + 18 = 9$$


$$f(5) = -25 + 30 = 5$$

Conclusión

El valor óptimo es:


$$x=3$$

y el beneficio máximo es:


$$f(3) = 9$$


La siguiente gráfica muestra la función, la restricción el valor optimo y el beneficio máximo.


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 5, 100)
y = -x**2 + 6*x

plt.plot(x, y, label="f(x) = -x² + 6x")
plt.scatter([3], [-3**2 + 6*3], color='red', label="Máximo")
plt.title("Optimización con restricción")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.legend()
plt.grid()
plt.show()
```

# Problema

Se desea construir un terreno rectangular con un **perímetro fijo de 100 metros**.  ¿Cuáles deben ser las dimensiones del rectángulo para **maximizar el área**?


**Planteamiento**

Sean:

$x$: largo del rectángulo

$y$: ancho del rectángulo


**Función objetivo (área)**:


$$A(x,y)=xy$$


**Restricción (perímetro)**:


$$2x + 2y = 100$$



**Reducción a una función objetivo de una variable**

Despejamos $y$ de la restricción:


$$2x + 2y = 100$$

$$y = 50 - x$$


Sustituimos en la función objetivo:


$$A(x) = x(50 - x)$$


$$A(x) = 50x - x^2$$


**Solución Analítica**

Derivada de la función objetivo en una variable


$$A'(x) = 50 - 2x$$



Punto crítico


$$50 - 2x = 0$$

$$x = 25$$

Se obtiene una de las dimensiones, el largo del rectángulo.


Se halla la dimensión restante


$$y = 50 - 25 = 25$$


**Conclusión**

Las dimensiones óptimas son:


$x = 25m$,  $y = 25m$


El área máxima es:


$$A(x,y)=xy$$


$$A(x,y) = (25m)(25m)= 625m^2$$


El rectángulo de mayor área con perímetro fijo es en realidad un **cuadrado**.



```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 50, 100)
y = x * (50 - x)

plt.plot(x, y, label="A(x) = x(50 - x)")
plt.scatter([25], [25*25], color='red', label="Máximo")
plt.title("Área máxima con perímetro fijo")
plt.xlabel("x (largo)")
plt.ylabel("Área")
plt.legend()
plt.grid()
plt.show()
```
