# **Introducción**

En cálculo diferencial, la **derivación implícita** es una técnica que se utiliza para derivar funciones que no están explícitamente resueltas en términos de una sola variable. Es decir, cuando una función no se encuentra despejada como


$$
y=f(x),
$$



sino que está dada mediante una relación entre $x$ e $y$, por ejemplo,


$$
x^2+xy+y^2=25.
$$



En estos casos puede no ser posible (o conveniente) despejar $y$. Por ello, se utiliza la **derivación implícita**, aplicando la **regla de la cadena** al derivar con respecto a $x$.



## **Procedimiento General**

Para derivar implícitamente:

1. Deriva ambos lados de la ecuación con respecto a (x).

2. Aplica la regla de la cadena a todos los términos que contienen (y):


$$
\frac{d}{dx}\left(y^n\right)
$$

$$
ny^{,n-1}\frac{dy}{dx}.
$$



3. Reúne todos los términos que contienen $\frac{dy}{dx}$.

4. Despeja $\frac{dy}{dx}$.



## Ejemplo 1

Derivar implícitamente


$$
x^2+y^2=25
$$



## Solución

Derivando ambos lados,


$$
\frac{d}{dx}(x^2)+\frac{d}{dx}(y^2)
$$



Se obtiene

$$
2x+2y\frac{dy}{dx}=0.
$$

Despejando,

$$
\frac{dy}{dx}=-\frac{x}{y}.
$$



## Ejemplo 2
Derivar implícitamente


$$
xy+y^2=10
$$



## Solución
Derivando ambos lados,


$$
\frac{d}{dx}(xy)+\frac{d}{dx}(y^2)=0.
$$



Aplicando la regla del producto,


$$
x\frac{dy}{dx}+y+2y\frac{dy}{dx}=0.
$$



Agrupando los términos,

$$
(x+2y)\frac{dy}{dx}=-y.
$$

Por lo tanto,

$$
\frac{dy}{dx}=-\frac{y}{x+2y}.
$$



## Ejemplo 3

Derivar implícitamente,

$$
\sin(xy)=x
$$



## Solución
Derivando,

$$
\cos(xy)\frac{d}{dx}(xy)=1.
$$

Como

$$
\frac{d}{dx}(xy)=x\frac{dy}{dx}+y,
$$



se obtiene

$$
\cos(xy)\left(x\frac{dy}{dx}+y\right)=1.
$$

Entonces,

$$
x\frac{dy}{dx}+y=\frac{1}{\cos(xy)}.
$$



Finalmente,

$$
\frac{dy}{dx}=\frac{1}{x\cos(xy)}-\frac{y}{x}
$$



## Ejemplo 4
Derivar implícitamente

$$
e^{xy}=x+y
$$



## Solución
Derivando ambos lados,

$$
\frac{d}{dx}(e^{xy})=\frac{d}{dx}(x+y).
$$



Aplicando la regla de la cadena,


$$
e^{xy}\left(x\frac{dy}{dx}+y\right)=1+\frac{dy}{dx}.
$$



Desarrollando,

$$
xe^{xy}\frac{dy}{dx}+ye^{xy}=1+\frac{dy}{dx}.
$$



Agrupando los términos con la derivada,

$$
xe^{xy}\frac{dy}{dx}-\frac{dy}{dx}=1-ye^{xy}.
$$

Factorizando,

$$
\left(xe^{xy}-1\right)\frac{dy}{dx}=1-ye^{xy}.
$$

Finalmente,

$$
\frac{dy}{dx}=\frac{1-ye^{xy}}{xe^{xy}-1}.
$$



## Ejemplo 5

Derivar implícitamente

$$
x^3+y^3=6xy
$$



## Solución
Derivando,

$$
3x^2+3y^2\frac{dy}{dx}=6\left(x\frac{dy}{dx}+y\right).
$$

Desarrollando,

$$
3x^2+3y^2\frac{dy}{dx}=6x\frac{dy}{dx}+6y.
$$

Reuniendo los términos con la derivada,

$$
3y^2\frac{dy}{dx}-6x\frac{dy}{dx}=6y-3x^2.
$$

Factorizando,

$$
(3y^2-6x)\frac{dy}{dx}=6y-3x^2.
$$

Finalmente,

$$
\frac{dy}{dx}=\frac{6y-3x^2}{3y^2-6x}.
$$


Que puede simplificarse dividiendo numerador y denominador entre 3:


$$
\frac{dy}{dx}=\frac{2y-x^2}{y^2-2x}.
$$
