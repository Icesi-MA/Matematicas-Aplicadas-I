**Notación**

Sea $𝑓$ una función definida y acotada en un intervalo $[𝑎, 𝑏]$. Esto es, asumimos que existen números $𝑚$ y $𝑀$ tales que

$$𝑚 ≤ 𝑓(𝑥) ≤ 𝑀$$  

para todo $𝑥$ en $[𝑎, 𝑏]$.


Sea el conjunto $P=\{x_0<x_1<x_2<...<x_n\}$ una partición del intervalo $[a,b]$.


Si

$$ a=x_0<x_1<x_2<...<x_n=b$$


entonces tal partición $P$ divide $[a,b]$ en $n$ intervalos $[x_{i-1},x_i]$ de longitud


$$ \Delta x_i= x_i-x_{i-1} $$


Para cada intervalo $[x_{i-1},x_i]$, sea $𝑀_𝑖$ el número más pequeño
tal que $𝑓(𝑥) ≤ 𝑀_𝑖$ para todo $𝑥$ en $[𝑥_{𝑖−1}, 𝑥_𝑖]$ y sea $𝑚_𝑖$ el número mayor tal que $𝑓(𝑥) ≥ 𝑚_𝑖$ para todo $𝑥$ en $[𝑥_{𝑖−1}, 𝑥_𝑖]$.





Si $𝑓(𝑥) ≥ 0$ para todo $𝑥$ en $[𝑥_{𝑖−1}, 𝑥_𝑖]$, entonces, en el lenguaje de nuestro análisis anterior, el rectángulo con base $[𝑥_{𝑖−1}, 𝑥_𝑖]$ y altura $𝑀_𝑖$ es un rectángulo circunscrito y el rectángulo de base $[𝑥_{𝑖−1}, 𝑥_𝑖]$ y altura $𝑚_𝑖$ es un rectángulo inscrito.

Ahora definimos

$$ U(f,P) = M_1\Delta x_1 + M_2\Delta x_2 +...+M_n\Delta x_n = \sum_{i=1}^{n} M_i\Delta x_i $$

la suma superior de 𝑓 con respecto a la partición 𝑃, y

$$ L(f,P) = m_1\Delta x_1 + m_2\Delta x_2 = +...+m_n\Delta x_n = \sum_{i=1}^{n} m_i\Delta x_i $$

la suma inferior de 𝑓 con respecto a la partición 𝑃.


Tenga en cuenta que siempre tenemos

$$  L(f,P) \leq U(f,P) $$


si 𝐴 es el área debajo de la gráfica de 𝑓 y encima del intervalo
[𝑎, 𝑏], esperaríamos poder hacer 𝑈(𝑓, 𝑃) y 𝐿(𝑓, 𝑃) arbitrariamente cerca de 𝐴. Esto implicaría que 𝐴 es el único número con la propiedad:

$$  L(f,P) \leq A \leq U(f,P) $$

Para toda particion $P$.


El número $A$ se llama la integral definida de la función en el intervalo considerado; como se vera en la siguiente definición.

**Deinición**

Una función 𝑓 es integrable en un intervalo [𝑎, 𝑏] si existe un número único 𝐼 tal que

$$  L(f,P) \leq I \leq U(f,P) $$

para todas las particiones 𝑃 de [𝑎, 𝑏].


Si 𝑓 es integrable en [𝑎, 𝑏], llamamos al número 𝐼 la integral definida de 𝑓 sobre [𝑎, 𝑏], y lo denotamos

$$ I=\int_{a}^{b} f(x) \,dx $$
