  1.3 Utilizando el método de Euler que vimos en clase, cuyo código en python podría ser el que encuentran en la siguiente celda, muestre primero analíticamente que si algún $x_n$ es igual a $a$, entonces $x_{n + 1} = a$, y haga lo mismo numéricamente asignando un valor arbitrario a la constante $a$.
  
  1.4 Considere un método numérico, similar al método de Euler,  utilizando la siguiente aproximación, $$ x_{n+1}=x_n-h(a-x_{n+1})$$
Explique en que se diferencia esta aproximación de la del método de Euler. De ser posible encuentre al menos un valor de $h$ tal que la secuencia $x_n$ alterne entre números menores que $a$ y mayores que $a$. Compruebe que su elección es correcta mostrando los resultados numéricamente.  


```python
import numpy as np
import math 
import matplotlib.pyplot as plt

def euler(f,x0,t0,tmax,n):
    t = np.linspace(t0, tmax, n)# domino de la solución, depende de t0, tmax, y n
    h = (tmax-t0)/n
    x = np.zeros_like(t) # Arreglo de ceros del mismo tamaño de t
    x[0] = x0# Condición inicial 
    for i in range( n-1 ):
        x[i+1] = x[i]+h*f(x[i],t[i]) # Aproximación de la  solución en cada instante t
    return t, x
```

```python
def rungekutta2(f,x0,t0,tmax,n):
  t = np.linspace(t0, tmax, n)# domino de la solución, depende de t0, tmax, y n
  h = (tmax-t0)/n
  x = np.zeros_like(t) # Arreglo de ceros del mismo tamaño de t
  x[0] = x0# Condición inicial 
  for i in range(n - 1):
      x[i+1] = x[i] + h * f(x[i] + (h/2.0)*f(x[i], t[i]) , t[i] + h/2.0)
  return t,x
```

```python
def cooling_rate(temperature, time, k=-0.34, ambient_temperature=25):
    return k * (temperature - ambient_temperature)

x, y = rungekutta2(cooling_rate, 95, 0, 10, 100)
u, v = euler(cooling_rate, 95, 0, 10, 100)
plt.plot(x, y, "b.-", label="Runge-Kutta 2")
plt.plot(u, v, "g.-", label="Euler")
plt.legend(loc="best")
plt.xlabel("t")
plt.grid()
plt.show()
```
