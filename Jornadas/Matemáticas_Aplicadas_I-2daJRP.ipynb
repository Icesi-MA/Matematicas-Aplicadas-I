# Modelando la realidad: resolviendo ecuaciones no lineales

A continuación vamos a estudiar uno  de los métodos de solución de ecuaciones no lineales $F(x)=0$, y de problemas de optimización por lo tanto, mas eficiente y popular en algunas aplicaciones: el método de *Newton* o *Newton-Rhapson*. Una ecuación de la forma $𝑓 (𝑥) = 0 $ puede resolverse para $𝑥$ mediante álgebra simple si $𝑓$ es una función afín y por la fórmula cuadrática si $𝑓$ es un polinomio cuadrático. Hay fórmulas similares a la fórmula cuadrática tanto para polinomios cúbicos como para polinomios cuarticos, pero son, en general, muy complicadas. Uno de los mas interesantes resultados de las matemáticas, debido a Niels Henrik Abel (1802-1829), es que 
no existe un análogo de la fórmula cuadrática para polinomios quínticos. Por esta y otras razones, resulta que en muchas situaciones resolver una ecuación $𝑓 (𝑥) = 0$ para $𝑥$ requiere del uso de un método que pueda aproximar las soluciones hasta un cierto
nivel de precisión. El método de Newton es un algoritmo muy sencillo cuyo código en python podría ser el que encuentran en la siguiente celda  


```python
def newton(f,Df,x0,epsilon,max_iter):
    ''' Solucion de f(x)=0 por el método de Newton.

    Parametros de entrada
    ----------
    f : funcion
    Df : Derivada de f
    x0 : Valor inicial 
    epsilon : Criterio de parada  abs(f(x)) < epsilon (precisión)
    max_iter : Maximum número de iteraciones  

    Valores de salida
    -------
    xn : Valor aproximado de la solución de la ecuación f(x)=0

    Ejemplo
    --------
    f = lambda x: x**2 - x - 1
    Df = lambda x: 2*x - 1
    newton(f,Df,1,1e-8,10)

    > Solución encontrada luego de 5 iteraciones.
    1.618033988749989
    '''
    xn = x0
    for i in range(0,max_iter):
        fxn = f(xn)
        if abs(fxn) < epsilon:
            print('Solución encontrada luego de',i,'iteraciones.')
            return xn
        Dfxn = Df(xn)
        if Dfxn == 0:
            print('Derivada cero. El método falla en éste punto, no puede encontrar la solución')
            return None
        xn = xn - fxn/Dfxn
        print(xn)
    print('Máximo número de iteraciones alcanzadas. No se encontró la solución.')
    return None
```
